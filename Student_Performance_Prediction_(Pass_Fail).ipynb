
Project Title: Student Performance Prediction

Description:
This project predicts whether a student will pass or fail based on study hours, attendance, and previous marks using Machine Learning.

Technologies Used:
- Python
- Pandas
- Scikit-learn
- Matplotlib

Model Used:
- Logistic Regression

```python
import pandas as pd

data = {
    "study_hours": [5, 2, 7, 3, 6],
    "attendance": [80, 50, 90, 60, 85],
    "previous_marks": [70, 40, 85, 50, 78],
    "result": ["Pass", "Fail", "Pass", "Fail", "Pass"]
}

df = pd.DataFrame(data)
df
```

```python
df.info()
df.isnull().sum()
```

```python
df["result"] = df["result"].map({"Pass": 1, "Fail": 0})
df
```

```python
X = df[["study_hours", "attendance", "previous_marks"]]
y = df["result"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
print(y_pred)
```

```python
from sklearn.metrics import accuracy_score

print(accuracy_score(y_test, y_pred))
```
